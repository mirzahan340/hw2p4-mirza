## Task 3: Handle a missing-value issue

Identify one column with missing values and decide how to handle it. You may:

### My missing-value strategy
the "age" column has 177 missing values (out of 891). I will fill these with the median age instead of the mean since age distributions are often biased towards the right. I chose to fill rather than drop the rows, since the amount of rows with missing ages is around 20 percent of the database. ;Filling instead of dropping will still incluyde every passenger.

```python


df['Age'] = df['Age'].fillna(df['Age'].median())

print("missing values after handling-")
print(df.isnull().sum())
```

### Why I grouped/filtered by this column

I grouped by passenger class (Pclass) and computed the mean of the surviving amount of people for each group. since "survived" is coded as 1, the mean will give us the survival rate. I wanted to see if class differences changed the survival rate.

```python


result = df.groupby('Pclass')['Survived'].mean()
print(result)
```
